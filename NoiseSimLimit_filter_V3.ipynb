# Simulate the signal and noise, then set limit with noise power.

We will study the first three moments of the noise distribution and use the to set limits. The constraning power scales as $E^1$, $E^2$ and $E^3$, respectively. So we expect to see the limit being extended far beyond the reach from a triggering analysis, and that the maximum restricting power transitions from trigger to third moments to second then to the first moment sequentially as the DM mass reduces and the spectrum being compressed to low energies. 

First, we will construct the formulas to discribe the signal spectral power for the 3 moments. Note that this can be done both in frequency domain and in the time domain. In frequency domain, the first moment is the uncorrelated noise PSD, the second moment is the correlated noise in a two channel device, the third moment is not explicit and we will try to find the definition. In time domain, the measureable is the energy spectrum of random events. First moment is the DC offset, which is not well-defined. The second moment is the variance. The third moment is the skew, and specifically, if normalize the skew by variance, we get skewness, which will be a powerfull tool to limit the high energy events.

# Setting limits from PSD.

There are two aspects in limit setting. We usually set limits base on triggered events. But actually, we can set limits from the noise power as well. The former aspect is only the ideal case where the time region of interest can be set around the signal peaks to boost the ratio of the signal energy integral to the noise energy integral. When the signals are small but constantly arriving, limiting the signal time stream to a certain period does not boost SNR anymore, because the same amount of signal power is contained in all time periods. We are in this regim when the expected signal rate is comparable to the signal bandwidth.

If we take random traces from continuous signal recordings, calculate the signal power referenced noise $S_{P}(\omega)$, and assume all measured power is contributed by dark matter signal, we can set the upper limit on dark matter rate.

## Single channel (First moment)
Assume the detector has power to current responsivity $s_{I}(\omega)$, equivalently the so-called $dI/dP(\omega)$. From K. Irwin, $s_{I}$ writes as 
\begin{equation}
    s_{I}(\omega) = -\left[ I_0 R_0 \left(
                           \frac{L}{\tau_{el}R_0\mathcal{L}_I} 
                           +(1-\frac{R_L}{R_0})
                           +\frac{i\omega L\tau}{R_0\mathcal{L}_I}(\frac{1}{\tau_I}+\frac{1}{\tau_{el}})
                           -\frac{\omega^2\tau}{\mathcal{L}_I}\frac{L}{R_0}
                           \right) \right]^{-1}
\end{equation}
where $I_0$ and $R_0$ are the steady state bias current and resistance of the TES, respectively. $L$ is the total inductance in the TES loop, $R_L$ is the load resistance in the loop. $\tau$, $\tau_{el}$ and $\tau_I$ are the natural, constant-current and electrical time constants, respectively. $\mathcal{L}_I$ is the TES loop gain. In our case, the signal is caused by athermal phonons, which has a characteristic phonon power pulse shape $p_s(t)$. The signal template $i_s(t)$ is a convolution of $p_s(t)$ and $\mathcal{F}^{-1}\{s_{I}(\omega)\}$.

The measured power-referenced power spectrum $S_p(\omega)$ is a combination of signal power, $P_s(\omega)$, and noise power, $P_n=\frac{n_I}{s_I}$. 
\begin{align}
    S_p(\omega) &=& \frac{S_I(\omega)}{|s_I(\omega)|^2} \\
                &=& \frac{1}{|s_I(\omega)|^2} [P_s(\omega) s_I(\omega) + n_I(\omega)][P_s^*(\omega) s_I^*(\omega) + n_I^*(\omega)]
\end{align}

To simplify the problem, we assume the athermal phonon pulse shapes are identical for low energy events. Then, we can define $p_0(t)$ as the unit-energy pulse in time domain, and $P_0(\omega) = \mathcal{F}\{p_0(t)\}$ in frequency domain. Note that we choose the orthorgnal normalization convention for Fourier transform on a finit time interval of length $T$ such that $|\mathcal{F}\{f(t)\}|^2$ represent the power spectral density (PSD). Under this convention, $P_0(\omega)\propto T^{-1/2}$. For events sampled from a spectrum $dR/dE$ of a certain model, $p_s(t)$ is the summation of $p_0(t)$ scaled to each event energy $E_i$ and delayed to $t_i$. In frequency domain, the time delay become random phases $\theta_i$, and we have 
\begin{equation}
    P_s(\omega) = P_0(\omega)\sum_{i=0}^N E_i e^{i\theta_i}
\end{equation}
where $N$ is the number of events on the trace of length $T$.

Now $S_p(\omega)$ can be write as 
\begin{align}
    S_p(\omega) &=& [P_0(\omega)\sum_{i=0}^N E_i e^{i\theta_i} + P_n(\omega)][P_0^*(\omega)\sum_{i=0}^N E_i e^{-i\theta_i} + P_n^*(\omega)] 
\end{align}

And the average power spectrum over many traces is
\begin{align}
    <S_p(\omega)> &=& <\sum E_i^2 P_0(\omega)P_0^*(\omega) + P_n(\omega)P_n^*(\omega)> \\
                & & + <\sum E_i E_j e^{i(\theta_i+\theta_j)} P_0(\omega)P_0^*(\omega) + \sum E_i e^{i\theta_i} P_0(\omega)P_n^*(\omega) + C.C.> \\
                &=& <\sum E_i^2> P_0(\omega)P_0^*(\omega) + P_n(\omega)P_n^*(\omega) \\
                &=& \left[T \int_{E_{min}} \frac{dR}{dE} E^2 dE \right] |P_0(\omega)|^2 + |P_n(\omega)|^2
\end{align}
Summations in line two averages to zero because of the random phases, only the self-correlated term of each event and the noise survive the time-average. This makes sense, because of the random phases, no interference happens between different events and the noise, the total power is a linear summation of the signal energies and the noise power. 

We assume the DM model yields signal 
\begin{equation}
    \frac{dR}{dE}(\sigma_0, m_{det}, m_{DM})
\end{equation}
in a detector with target mass $m_{det}$ at the reference cross section $\sigma_0$. The summation of event rate begines from $E_{min}$, which is the superconducting bandgap of Al, $E_{min} = 2\Delta = 3.5 k_B T_c \approx 0.36 meV$.
In the end, the signal powers scales with trace length $T$. This is because the normalization of the PSD of a pulse event has the term $1/T$, i.e. the energy of the signal template is considered to be spread over the entire trace $T$, and $|P_0(\omega)|^2$ scales as $1/T$. $T$ from the total event number and $1/T$ from signal pulse PSD cancel each other. On the other hand, the noise power is continuous and $|P_n(\omega)|^2$ is independent of $T$.

From this equation, we can derive a simple limit-setting scheme: 
1) Acquire **random** traces **without large pulses**. As lone as the expected number of events in time interval $T$ at the final cross section limit is much greater than one, the selection of randoms should not affect the result. In other words, cut efficiency calculation is not necessary. 

2) Calculate the PSD of the traces, calculate the average PSD.
3) Use $dI/dV$ data to measure and calculate $dI/dP$ and convert the current PSD to power-referenced PSD $<S_p(\omega)>$.

4) Acquire triggers with optimum filter, build event template, deconvolve the TES response and build the phonon pulse template $p_0(t)$. The energy collection efficiency should be included into the normalization. Then Fourier transform to $P_0(\omega)$

5) The upper limit of $\left[T \int_{E_{min}} \frac{dR}{dE} E^2 dE \right]$ is $\min_{\omega}(<S_p(\omega)>)$. Because the average PSD have sufficient exposure, the confidence interval should be limited by the systematic uncertainties in pulse shape. 

6) Depending on the choice of DM model, extrace the cross section limit. 

## Double channel (Second moment)

This method is more powerful in a double-channel device. Phonons from bulk events will distribute almost evenly in the two channels, creating correlated signals. Phonons from singles induced by micro-fractures will concentrate almost in a single channel. 

We can calculate the correlated spectral density (CSD) of two channels
\begin{align}
    CSD_{12}(\omega) &\equiv& \mathcal{F}\left\{\lim_{T\to\infty}\frac{1}{T}\int_0^T f_2(t)f_1(t-\tau) d\tau \right\} \\
                     & = & \mathcal{F}\{f_1(t)\}^*\mathcal{F}\{f_2(t)\} \\
                     & = & F_1^*(\omega) F_2(\omega)
\end{align}
Unlike PSD, CSD is complex by definition, and the imaginary term represents the out-of-phase quadrature. We will set limits on the real part of the CSD, commonly referred to as cospectrum. 

Assume we have signals $E_i$, independent signal events $E_{s1i}, E_{s2i}$, and independent noises in the two channels. The power-referenced signals are
\begin{align}
    P_1(\omega) &=& \left(\sum \frac{1}{2}E_i e^{i\theta_i} + \sum E_{s1i} e^{i\theta_{s1i}}\right)P_0(\omega) + P_{n1}(\omega) \\
    P_2(\omega) &=& \left(\sum \frac{1}{2}E_i e^{i\theta_i} + \sum E_{s2i} e^{i\theta_{s2i}}\right)P_0(\omega) + P_{n2}(\omega) \\
\end{align}

First, without loss of generality, let's simplify the summations of correlate signals and singles to one event each
\begin{align}
    P_1(\omega) &=& \left(\frac{1}{2}E e^{i\theta} + E_{s1} e^{i\theta_{s1}}\right)P_0(\omega) + P_{n1}(\omega) \\
    P_2(\omega) &=& \left(\frac{1}{2}E e^{i\theta} + E_{s2} e^{i\theta_{s2}}\right)P_0(\omega) + P_{n2}(\omega) \\
\end{align}

And the power-referenced CSD is
\begin{align}
    P_1^*(\omega) P_2(\omega) &=& P_0^*(\omega) P_0(\omega) \left(\frac{1}{4}E^2 
                                    + \frac{1}{2}E(E_{s1}e^{-i(\theta_{s1}-\theta)} + E_{s2}e^{-i(\theta-\theta_{s2})}) 
                                    + E_{s1}E_{s2}e^{-i(\theta_{s1}-\theta_{s2})} \right) \\ 
                                  & & + P_{n1}^*(\omega) P_0(\omega)\left(\frac{1}{2}E e^{i\theta} + E_{s2} e^{i\theta_{s2}}\right) + P_0^*(\omega) P_{n2}(\omega) \left(\frac{1}{2}E e^{-i\theta} + E_{s1} e^{-i\theta_{s1}}\right)\\
                                  & & + P_{n1}^*(\omega) P_{n2}(\omega)
\end{align}
All terms besides the first coherent $\frac{1}{4}V_s(\omega)^*V_s(\omega)E^2$ term in the first line have random phases. They will average to zero after taking the real part and average over multiple traces. 
\begin{equation}
 <Re(CSD_{12})> = \frac{1}{4}E^2 |P_0(\omega)|^2
\end{equation}
No surprises, the noise terms disappear after averaging. However, absolute zero will not be achieved in a real measurement. The mean value converges to zero as $1/\sqrt{N}$ where $N$ is the total number of traces.

In general, when we have multiple signal hits and possibly correlated noise powers on the two channels, the average cospectrum generalizes to 
\begin{equation}
<Re(CSD_{12})> = \frac{T}{4}\int_{E_{min}}\frac{dR}{dE}E^2dE |P_0(\omega)|^2 + |P_{n corr}(\omega)|^2
\end{equation}
Comparing to the case of a single channel, there is an additional factor of $1/4$. This is because of the energy partition between the two channels. Assuming an even-partition yields conservative limits as $a(1-a)\leq 1/4$ for $a\in[0,1]$. Same as the single-channel case, the dark matter limit can be set by 
\begin{equation}
T\int_{E_{min}}\frac{dR}{dE}E^2dE \leq 4\min_{\omega}\left(<Re(CSD_{12}(\omega))>\right)
\end{equation}

## Third moment -- a unified notation

A device with three channels is less efficient. Instead, we will first check what the 3rd moment looks like in a single channel device. 

First, assume the trace has signal only,
\begin{equation}
Y(\omega) = \sum Y_m \widetilde{T}(\omega)e^{-i\omega t_m}
\end{equation}
$Y_m$ is the energy of the $m$th event, and $t_m$ is its time offset. We assume the events follows Poission distribution.

The first moment $\langle Y(\omega) \rangle_T = R\langle Y\rangle \delta(\omega)$ is triviel, it is the DC offset multipled by the Dirac delta function $\delta(\omega)$, meaning all AC components are expected to be zero after the averaging. In time domain $\langle Y(T) \rangle_N=R\langle Y\rangle$ is just average template height times the total event rate. In other words, the DC offset of the baseline given the signal rate. 

The second moment, 
\begin{equation}
\langle \Delta Y(\omega_1) \Delta Y^*(\omega_2) \rangle_T = \widetilde{T}(\omega_1)\widetilde{T}^*(\omega_2)R\langle Y^2\rangle 2\pi \delta(\omega_1-\omega_2)
\end{equation}
Same as the derived equations above. When we consider $Y_1$ and $Y_2$ from two channels, this is the correlated noise. We we consider $Y_1$ and $Y_2$ of the same channel, this is the PSD. We can also integrate over the frequencies to get the time domain result:
\begin{equation}
\langle \Delta Y(t) \Delta Y(t) \rangle_N = \int_{-\infty}^\infty \frac{d\omega}{2\pi}|\widetilde{T}(\omega)|^2 R\langle Y^2\rangle 
\end{equation}
Which is just the variance of the spectrum.

Now, let's look at the third moment.
\begin{equation}
\langle \Delta Y(\omega_1) \Delta Y(\omega_2) \Delta Y(\omega_3)\rangle_T = \widetilde{T}(\omega_1)\widetilde{T}(\omega_2)\widetilde{T}(\omega_3)R\langle Y^3\rangle 2\pi \delta(\omega_1+\omega_2+\omega_3)
\end{equation}
Here, we do not take complex conjugates, so all three frequencies are symetrical, this results in the summation in the delta funciton. And we have in time domain:
\begin{equation}
\langle (Y(t)-\langle Y(t)\rangle)^3 \rangle_N = \int\int \frac{d\omega_1d\omega_2}{4\pi^2}\widetilde{T}(\omega_1)\widetilde{T}(\omega_2)\widetilde{T}^*(\omega_1+\omega_2) R\langle Y^3\rangle 
\end{equation}



No we can understande the relation between frequency and time domain:
1. They scale to the same moment of the amplitude distribution, $\langle Y^N \rangle$
2. They have the same linearly scaling with total rate $R$
3. In frequency domain, there is extra frequency dependence, which is 0,1,2 dimentional for the first, second and third moments. But in time domain, the dependece is integrated out to be a single constant factor. The factor represent how easily the signals pile-up together. If $T(t)$ is a delta function, $\widetilde{T}(\omega)$ is a flat distribution. The prefactors are all unit constants. For any finit $T(t)$ width, $\widetilde{T}(\omega)$ will be frequency dependent, and the integration of the time domain prefactor will reduce the limiting power as frequency ranges where $\widetilde{T}(\omega)$ is small will be included. For example, in the extreme case, when $\widetilde{T}(\omega)$ is a delta function, the limiting power of the frequency domain is infinite. A delta function at zero frequency is better approximated with a slower detector, meaning, reducing the detector bandwidth can improve the limit power in frequency domain.


## Statistic
Now that we know the behaviour of the mean expectations, we should look at the distributions and design a good statistics.

Let's look at a specific frequency component $\omega$ of the spectrum. Assume there is no background noise, only signals of the known template $P_0(\omega)$ add up with random phases. So the total amplitude vector $P_s(\omega) = \sum_i E_i e^{i\theta_i} P_0(\omega)$ is an implementation of a 2D random walk on the complex plane, whose magnitude follows a Gaussian distribution under the central limit theorem  
\begin{equation}
|P_s(\omega)| \sim\frac{1}{\sqrt{2\pi\sigma^2}} \exp(-\frac{|P_s(\omega)|^2}{2\sigma^2})
\end{equation}
where
\begin{equation}
\sigma^2 = T \int_{E_{min}}\frac{dR}{dE}E^2dE |P_0(\omega)|^2
\end{equation}
And the power-referenced PSD at frequency $\omega$ follows an exponential distribution
\begin{equation}
S_p(\omega)=|P_s(\omega)|^2 \sim \frac{1}{\sigma^2} \exp(-\frac{|P_s(\omega)|^2}{\sigma^2})
\end{equation}
We can use the normalized vairable $|P_s(\omega)|^2/|P_0(\omega)|^2$ to remove the frequency dependence
\begin{equation}
X \equiv \frac{|P_s(\omega)|^2}{|P_0(\omega)|^2} \sim \frac{1}{\tau_0} \exp(-\frac{X}{\tau_0})
\end{equation}
where $\tau_0 = T \int_{E_{min}}\frac{dR}{dE}E^2dE$ is the same for all frequency bins.
When there is noise power $P_n(\omega)$ added randomly to the signal power, the exponential $\tau_0$ can be generalized as
\begin{equation}
\tau(\omega) = \tau_0+|P_n(\omega)|^2/|P_0(\omega)|^2
\end{equation}
When there is noise at a certain frequency, for example, the 60Hz power line pick-up, we need to revise the random walk picture. This noise frequency is a large vector that offsets the random walk of other small vectors from the origin, and the distribution of $P(\omega)$ is no longer exponential. However, as we are looking for the upper limits, we want to look for the frequency bin with the smallest $\tau(\omega)$, and set the limit with $\tau_0\leq\tau$. Ignoring the frequencies with distributions that diverge from an exponential is safe. 

Now, the limit-setting simplifies to this problem: At each frequency $\omega$, the random variable $X\sim Exp(\tau)$ is sampled $N$ times, and this experiment is repeated for $M$ frequencies. What is the upper limit of $\tau_0\leq \min_{\omega}\hat{\tau}$ ?

When $N\gg 1$, we can simply estimate $\tau$ from the mean values $\hat{\tau}=<X>$, look for $\min_\omega<X>$, and correct for the look-elsewhere effect of looking for the minimum in $M$ independent experiments. $P_{<X>}(x|\tau)$ is the probability of $<X>=x$, and it is Gaussian.
\begin{align}
P_{<X>}(x|\tau) = \frac{1}{\sqrt{2\pi N}\tau} e^{-(x-\tau)^2/2\tau^2N}
\end{align}
Then $P_{\min_\omega <X>}(x|\tau)$ denotes the probability of the minimum $<X>$ in all frequencies to be $x$.
\begin{align}
P_{\min_\omega <X>}(x|\tau) & = & M P_{<X>}(x|\tau)\left(1-CDF_{<X>}(x|\tau)\right)^{M-1} \\
\end{align}
 The upper limit $\tau_{lim}$ at 90\% confidence level with a observed minimum $\min_\omega <X>=x_{min}$ is obtained by solving
\begin{equation}
CDF_{\min_\omega <X>}(x_{min}|\tau_{lim}) = 0.1
\end{equation}
and finally 
\begin{equation}
\sigma_{DM lim} = \frac{\sigma_0 \tau_{lim}}{T\int_{E_{min}}\frac{dR}{dE}(\sigma_0,m_{det},m_{DM})E^2dE}
\end{equation}

Note that this simple method is conservative in the large sample limit. 

1) We consider all noise power as signal power, resulting in overestimating the true signal power.

2) We assume the $M$ tests are drawn from the same exponential with the coefficient $\tau$, but the distributions may differ because of additional noise power. In this case, we want to set the upper limit of the minimum $\tau(\omega)$. If $\tau(\omega)\in[\tau_{min},\tau_{max}]$, it is obvious that $CDF_{\min_\omega <X>}(x_{min}|[\tau_{min},\tau_{max}])<CDF_{\min_\omega <X>}(x_{min}|\tau_{min})$. It means that our limit of $\tau$ under the frequency-independent assumption relates to a limit of the minimum $\tau(\omega)$ at a higher confidence level. So, the frequency-independent approximation is conservative.

3) We assume the $M$ tests are independent, which is true. Because the random phase is $\theta_i = \omega t_i$, where $t_i$ is the uniformly distributed time offset of each signal pulse $i$. With enough events on a single trace, $\omega t_i$ folds over $[0,2\pi]$ becomes a new uniform distribution for different $\omega$.

4) We should define the frequency range of the $M$ tests before calculating the noise PSD. The range can be limited to $dI/dP$ 3dB point at the high-frequency end and above 60Hz at the low-frequency end. Minimizing $M$ around the best SNR frequency with "educated guess" can avoid additional penalties from the look-elsewhere effect.

```python
# Functions to process traces

def generate_random_energy(x, Px, N):
    """
    Generate N random numbers according to a distribution represented by (x, P(x)),
    using linear interpolation between points.

    Parameters:
        x (numpy array): The values of the random variable.
        Px (numpy array): The corresponding probabilities.
        N (int): The number of random numbers to generate.

    Returns:
        numpy array: An array of N random numbers sampled from the distribution.
    """
    ## Normalize the probabilities (ensure they sum to 1)
    #Px_normalized = Px / np.trapz(Px,x)

    # Compute the cumulative distribution function (CDF)
    dx = np.diff(np.insert(x,0,0))
    cdf = np.cumsum(Px*dx)
    cdf /= cdf[-1]

    # Create a linear interpolation function for the CDF
    #cdf_interp = interp1d(x, cdf, kind='linear', fill_value=[0,1])# fill_value="extrapolate")

    # Generate N uniform random numbers between 0 and 1
    uniform_random = np.random.rand(int(N))

    # Map the uniform random numbers to the distribution using the inverse CDF
    # To find the inverse, we solve cdf_interp(x) = u for x
    random_numbers = np.array([np.interp(u, cdf, x) for u in uniform_random])

    return random_numbers


def generate_random_trace(E_list, t0_list, paddedtemp):
    """
    E_list: number of energy points on TWICE the trace length.
    Assum template are 20ms long with 1us binning. The template is padded with zeros of the trace length before and after.
    """
    trace = np.zeros(tracelength)
#t0_list = np.random.rand(len(E_list))*2*tracelength
    for i,E in enumerate(E_list):
        t0_bin = int(t0_list[i]) - tracelength*1
        trace += paddedtemp[tracelength-t0_bin:tracelength*2-t0_bin]*E
    return trace

def generate_random_trace_2ch(E_list, E_bkg_list, t0_list, t0_bkg_list, padded_temp):
    """
    E_list: number of energy points on TWICE the trace length.
    E_bkg_list: number of bkg events in individual channels on TWICE the trace length.
    PhononTau: Phonon collection time
    Assum template are 20ms long with 1us binning. The template is padded with zeros of the trace length before and after.
    padded_temp = [singles, convolved, phonon_number!]
    phonon template truncate at t where E_ph = 2DeltaAl.
    """
    padded_temp_full = padded_temp[1]
    padded_temp_single = padded_temp[0]
    phonon_temp = padded_temp[2]
    phonon_temp /= np.sum(phonon_temp)
    trace = np.zeros((2,tracelength))
    #trace_debug = np.zeros((2,tracelength))
#t0_list = np.random.rand(len(E_list))*2*tracelength
    for i,E in enumerate(E_list):
        t0_bin = int(t0_list[i]) - tracelength*1
        # if there are more than 100 phonons, use convolved template.
        #trace_debug += [padded_temp_full[tracelength-t0_bin:tracelength*2-t0_bin]*E*0.2, 
        #              padded_temp_full[tracelength-t0_bin:tracelength*2-t0_bin]*E*0.2]
        if E > E_ph_avg*100:
            
            E_left = np.random.binomial(n=int(E/E_ph_avg), p=0.2)*E_ph_avg
            E_right = np.random.binomial(n=int(E/E_ph_avg), p=0.2)*E_ph_avg
            trace += [padded_temp_full[tracelength-t0_bin:tracelength*2-t0_bin]*E_left, 
                      padded_temp_full[tracelength-t0_bin:tracelength*2-t0_bin]*E_right]
        else:
            # break energy into phonons
            # for each phonon:
            # first random a arriving time
            # then random an energy according to the time
            # go to next phonon, untill all energies are simulated.
            # if E_total > 100 * E_ph_avg, use convoluted pulse shape.
            N_ph_avg = int(E/E_ph_avg)
            N_ph_max = int(N_ph_avg + N_ph_avg**0.5*3 + 2)
            t_ph = np.random.choice(np.arange(len(phonon_temp))+1, size=N_ph_max, p=phonon_temp)
            ch_ph = np.random.choice([1,2,3], size=N_ph_max, p=[0.2,0.2,0.6])
            E_ph = 0.01*(t_ph*dt/3e-7)**-0.2 # eV From Matt's plot
            #print(E_ph)
            #print(f'total phonon energy is {np.sum(E_ph)}, total event energy is {E}')
            if np.sum(E_ph)<E :
                N_ph = N_ph_max
            else:
                N_ph = np.argwhere(np.cumsum(E_ph)>E)[0][0]
                if N_ph == 0:
                    E_ph = [E]
                    N_ph = 1
                elif N_ph==N_ph_max:
                    E_ph[-1] = E - np.sum(E_ph[:-1])
                else:
                    N_ph += 1
                    # adjust the energy of the last phonon
                    E_ph[N_ph-1] = E - np.sum(E_ph[:N_ph-1])
            #print(E_ph[:N_ph])
            #print(f'total phonon energy is {np.sum(E_ph[:N_ph])}, total event energy is {E}')  
            for iph in range(N_ph):
                if t0_bin+t_ph[iph]<=tracelength:
                    if ch_ph[iph] == 1: # left
                        #print(ch_ph[iph], t_ph[iph], t0_bin, E_ph[iph])
                        trace[0] +=  padded_temp_single[tracelength-t0_bin-t_ph[iph]:tracelength*2-t0_bin-t_ph[iph]]*E_ph[iph]
                    elif ch_ph[iph] == 2: # right
                        #print(ch_ph[iph], t_ph[iph], t0_bin, E_ph[iph])
                        trace[1] +=  padded_temp_single[tracelength-t0_bin-t_ph[iph]:tracelength*2-t0_bin-t_ph[iph]]*E_ph[iph]        
    #now simulate bkg
    # Shared bkg can be included in the signale channel!
    for i,E in enumerate(E_bkg_list):
        # singles are 100% efficiency in either left or right.
        t0_bin = int(t0_list[i]) - tracelength*1
        if np.random.rand()>0.5: # left
            trace[0]+=padded_temp_single[tracelength-t0_bin:tracelength*2-t0_bin]*E
        else:
            trace[1]+=padded_temp_single[tracelength-t0_bin:tracelength*2-t0_bin]*E
    #plt.plot(trace[0])
    #plt.plot(trace[1])
    #plt.plot(trace_debug[0],':')
    #plt.show()
    return trace

def generate_random_trace_fft(E_list, t0_list, temp_omega, freqs):
    trace_fft = np.zeros(tracelength,dtype=np.complex128)
    #t0_list = np.random.rand(len(E_list))*2*tracelength
    for i,E in enumerate(E_list):
        if np.abs(t0_list[i]-tracelength) <= tracelength//2:
            trace_fft += temp_omega * np.exp(-1j * (t0_list[i]- tracelength//2)*dt *2*np.pi* freqs) *E
    return trace_fft
    
#def downsizelog(value):
#    l=(len(value)//500)
#    value_short = np.concatenate((value[0:25*l], 
#                                np.mean(value[25*l:195*l].reshape((-1,10)), axis=-1),#.sum(axis=-1)/10.,
#                                np.mean(value[195*l:495*l].reshape((-1,100)), axis=-1)) )#.sum(axis=-1)/100.))
#                                #value[995*l:2995*l].reshape((-1,1000)).sum(axis=1)/1000.,))
#    return value_short
def downsizelog(value):
    l=(len(value)//500)
    value_short = np.concatenate((value[0:15*l], 
                                np.mean(value[15*l:95*l].reshape((-1,10)), axis=-1),#.sum(axis=-1)/10.,
                                np.mean(value[95*l:495*l].reshape((-1,100)), axis=-1)) )#.sum(axis=-1)/100.))
                                #value[995*l:2995*l].reshape((-1,1000)).sum(axis=1)/1000.,))
    return value_short
def downsizeloglog(matrix):
    processed_rows = np.array([downsizelog(row) for row in matrix])
    m_short = np.array([downsizelog(col) for col in processed_rows.T]).T
    #print(m_short)
    return m_short
```

```python
#Generate all energies, total points double to account for half-pulses outside the trace.

def generate_templates_2ch(taus, dt, tracelength):
    '''
    Generate templates for 1 or two channels
    taus = [tau_qp_r, tau_qp_f, tau_ph_r, tau_ph_f]
    nch = 1: single channel templates
    nch = 2: two channel templates + CSDs?
    '''
    freqs = fft.fftfreq(tracelength, d=dt)
    
    t_temp = np.arange(-dt*tracelength/2,dt*tracelength/2,dt)
    ph_temp = np.where(t_temp>0,1/(tau_ph_f*tau_ph_f/(tau_ph_f+tau_ph_r))*(1-np.exp(-t_temp/tau_ph_r))*np.exp(-t_temp/tau_ph_f),0)
    padded_phonon_temp = np.pad(ph_temp,(tracelength,tracelength),mode='constant',constant_values=(ph_temp[0],ph_temp[-1]))
    single_temp = np.where(t_temp>0,1/(tau_qp_f*tau_qp_f/(tau_qp_f+tau_qp_r))*(1-np.exp(-t_temp/tau_qp_r))*np.exp(-t_temp/tau_qp_f),0)
    padded_single_temp = np.pad(single_temp,(tracelength,tracelength),mode='constant',constant_values=(single_temp[0],single_temp[-1]))
    total_temp = np.convolve(ph_temp,single_temp,mode='same')*dt
    padded_total_temp = np.pad(tota_temp,(tracelength,tracelength),mode='constant',constant_values=(tota_temp[0],tota_temp[-1]))
    # phonon NUMBER (not energy) tempate tracelength//2+1 is the first non-zero bin
    phonon_N_temp = ph_temp[tracelength//2+1:]*(t_temp[tracelength//2+1:]/3e-7)**0.2
    #print('==========debug===========')
    #print(t_temp[tracelength//2+1:])
    #print(ph_temp[tracelength//2+1:])
    #print(phonon_N_temp)
    #plt.plot(phonon_N_temp)
    #plt.show()
    phonon_N_temp /= np.sum(phonon_N_temp)
    # templates for generate 2ch random trace with single phonon distributions.
    padded_temp = [padded_single_temp, padded_total_temp, phonon_N_temp]
    #print('==========debug===========')
    #print(phonon_N_temp)
    #plt.plot(phonon_N_temp)
    #plt.show()
    temps = [single_temp, total_temp]
    # PSD and skew matrix
    # Get the length of fft
    n_freqs = len(freqs)
    # Create a grid of indices
    i_fft, j_fft = np.meshgrid(np.arange(n_freqs//2), 
                               np.arange(n_freqs//2), indexing='ij')
    i_freq, j_freq = np.meshgrid(freqs[:n_freqs//2], 
                               freqs[:n_freqs//2], indexing='ij')
    #i_fft, j_fft = np.meshgrid(np.append(np.arange(0:n_freqs//2+1),np.arange(-n_freqs//2,0)) 
    #                           np.append(np.arange(0:n_freqs//2+1),np.arange(-n_freqs//2,0)), indexing='ij')
    # Compute i + j
    i_plus_j = i_fft + j_fft
    i_minus_j = i_fft - j_fft
    # Create a mask for valid indices (where i + j < n)
    valid_mask_plus = np.abs(i_plus_j) < n_freqs//2
    valid_mask_minus = np.abs(i_minus_j) < n_freqs//2
    i_plus_j = np.where(valid_mask_plus, i_plus_j, 0)
    i_minus_j = np.where(valid_mask_minus, i_minus_j, 0)
    # Analytical form
    
    omega = freqs*2*np.pi#np.arange(-tracelength//2,tracelength//2)/tracelength*1/dt*2*np.pi
    ##analytical_fft = lambda omega: np.exp(1j*omega*10e-3)/(tau_f*tau_f/(tau_r+tau_f))*( tau_f/(1+1j*omega*tau_f) - tau_r*tau_f/(tau_r+tau_f+1j*omega*tau_r*tau_f)) / (dt * tracelength)**0.5
    #analytical_fft = lambda omega: 1/(tau_f*tau_f/(tau_r+tau_f))*( tau_f/(1+1j*omega*tau_f) - tau_r*tau_f/(tau_r+tau_f+1j*omega*tau_r*tau_f)) / (dt * tracelength)**0.5
    #temp_omega = analytical_fft(omega)

    analytical_single_fft = lambda omega: np.exp(1j*omega*10e-3)*1/(tau_qp_f*tau_qp_f/(tau_qp_r+tau_qp_f))* \
                               ( tau_qp_f/(1+1j*omega*tau_qp_f) - tau_qp_r*tau_qp_f/(tau_qp_r+tau_qp_f+1j*omega*tau_qp_r*tau_qp_f)) /\
                               (dt * tracelength)**0.5
    analytical_total_fft = lambda omega: analytical_single_fft(omega) * 1/(tau_ph_f*tau_ph_f/(tau_ph_r+tau_ph_f))* \
                               ( tau_ph_f/(1+1j*omega*tau_ph_f) - tau_ph_r*tau_ph_f/(tau_ph_r+tau_ph_f+1j*omega*tau_ph_r*tau_ph_f)) 
    temps_omega = [analytical_single_fft(omega), analytical_total_fft(omega)]
    
    Skew_single_analytical = np.where(valid_mask_plus, analytical_single_fft(i_freq*2*np.pi) * \
                                                       analytical_single_fft(j_freq*2*np.pi) * \
                                                       analytical_single_fft(-(i_freq+j_freq)*2*np.pi), 1e-10)
    Skew_single_analytical += np.where(valid_mask_minus, analytical_single_fft(i_freq*2*np.pi) * \
                                                         analytical_single_fft(-j_freq*2*np.pi) * \
                                                         analytical_single_fft(-(i_freq-j_freq)*2*np.pi), 1e-10)
    Skew_single_analytical = np.array(np.real(Skew_single_analytical)*2/2/np.pi,dtype=float)##check why 2pi?
    Skew_total_analytical = np.where(valid_mask_plus, analytical_total_fft(i_freq*2*np.pi) * \
                                                      analytical_total_fft(j_freq*2*np.pi) * \
                                                      analytical_total_fft(-(i_freq+j_freq)*2*np.pi), 1e-10)
    Skew_total_analytical += np.where(valid_mask_minus, analytical_total_fft(i_freq*2*np.pi) * \
                                                        analytical_total_fft(-j_freq*2*np.pi) * \
                                                        analytical_total_fft(-(i_freq-j_freq)*2*np.pi), 1e-10)
    Skew_total_analytical = np.array(np.real(Skew_total_analytical)*2/2/np.pi,dtype=float)##check why 2pi?
    Skew_analytical = [Skew_single_analytical, Skew_total_analytical]
    
    # normalization
    OF_single_temp = np.convolve(single_temp, single_temp[::-1], mode='same')*dt
    OF_single_max = np.max(OF_single_temp - np.sum(OF_single_temp)/tracelength)
    OF_total_temp = np.convolve(tota_temp, tota_temp[::-1], mode='same')*dt
    OF_total_max = np.max(OF_total_temp - np.sum(OF_total_temp)/tracelength)
    OF_norm = [1/OF_single_max, 1/OF_total_max]
    return padded_temp, temps, temps_omega, Skew_analytical, OF_norm

def detect_pulses(data, threshold, time=None):
    """
    Detect pulses in a continuous data array and return:
    - Threshold-crossing times (rising edges)
    - Maximum values of each pulse
    - Timestamps of maxima

    Parameters:
        data (array): Input data array - OF filtered pulse 1D!!.
        time (array): Corresponding time array. If None, uses indices.
        threshold_factor (float): Threshold multiplier over baseline noise.

    Returns:
        list: List of dictionaries with pulse details.
    """
    # Robust baseline and noise estimation
    #baseline = np.median(data)
    #mad = np.median(np.abs(data - baseline))
    #rms = mad * 1.4826  # Convert MAD to RMS for Gaussian noise
    #threshold = baseline + threshold#_factor * rms

    # Find threshold crossings
    above_threshold = data > threshold
    edges = np.diff(above_threshold.astype(int))
    rise_idxs = np.where(edges == 1)[0]  # Rising edge indices
    fall_idxs = np.where(edges == -1)[0]  # Falling edge indices

    pulses = []
    #pulses_time=[]
    for i_rise in rise_idxs:
        # Calculate rising edge crossing time
        if i_rise + 1 >= len(data):
            continue
        y0 = data[i_rise] - threshold
        y1 = data[i_rise+1] - threshold
        if not (y0 <= 0 and y1 > 0):
            continue  # Skip invalid crossings
        
        ## Linear interpolation for precise crossing time
        #frac = -y0 / (y1 - y0)
        #if time is None:
        #    cross_time = i_rise + frac
        #else:
        #    if i_rise + 1 >= len(time):
        #        continue
        #    dt = time[i_rise+1] - time[i_rise]
        #    cross_time = time[i_rise] + frac * dt

        # Find corresponding falling edge
        j_fall = None
        fall_time = None
        if len(fall_idxs) > 0:
            idx = np.searchsorted(fall_idxs, i_rise, side='right')
            if idx < len(fall_idxs):
                j_fall = fall_idxs[idx]
                # Calculate falling edge crossing time
                if j_fall + 1 >= len(data):
                    continue
                y0_fall = data[j_fall] - threshold
                y1_fall = data[j_fall+1] - threshold
                if y0_fall >= 0 and y1_fall < 0:
                    frac_fall = -y0_fall / (y1_fall - y0_fall)
                    if time is None:
                        fall_time = j_fall + frac_fall
                    else:
                        if j_fall + 1 >= len(time):
                            continue
                        dt_fall = time[j_fall+1] - time[j_fall]
                        fall_time = time[j_fall] + frac_fall * dt_fall

        # Define pulse data window
        start_idx = i_rise + 1
        end_idx = j_fall + 1 if j_fall is not None else len(data)
        pulse_data = data[start_idx:end_idx]
        
        if len(pulse_data) == 0:
            continue  # Skip empty pulses

        # Find maximum and its time
        max_val = np.max(pulse_data)
        max_idx_rel = np.argmax(pulse_data)
        max_idx_abs = start_idx + max_idx_rel
        
        if time is not None:
            if max_idx_abs >= len(time):
                continue
            max_time = time[max_idx_abs]
        else:
            max_time = max_idx_abs

        pulses.append(max_val)
    #    pulses_time.append(max_time)
    #plt.plot(data) 
    #plt.scatter(pulses_time, pulses)
    #plt.plot([0,len(data)],[threshold,threshold])
    #plt.show()
    return pulses

def simulate_sig_and_noise_and_bkg(sigmaDM_true, DM_rates, bkg_rates, trace_par, taus, noise_PSD=[], pdf=None):
    '''
    noise_PSD should have the same length as the trace!
    taus = [tau_qp_r, tau_qp_f, tau_ph_r, tau_ph_f]
    trace_par = [dt, tracelength, N_trace]
    DM_rates = [mDM, EdRdE, init_rate]
    bkg_rates = [[beta_single, single_rate], [beta_share, share_rate]] #rate in Hz
    '''
    mDM = DM_rates[0]
    EdRdE = DM_rates[1]
    init_rate = DM_rates[2]
    dt = trace_par[0]
    tracelength = trace_par[1]
    N_trace = trace_par[2]
    tau_qp_r, tau_qp_f, tau_ph_r, tau_ph_f = taus
    
    freqs = fft.fftfreq(tracelength, d=dt)
    
    #create templates
    padded_temp, temp, temp_omega, Skew_analytical, OFamptoE = generate_templates_2ch(taus, dt, tracelength)
    #norm_PtoE = 1/np.max(np.convolve(temp, temp[::-1], mode='same')*dt)
    print(f'OFamptoE = {OFamptoE[0]}(single), {OFamptoE[1]}(shared)')
    
    #create random energies for DM signal
    total_rate = np.trapz(init_rate, EdRdE)
    E_mean = np.trapz(init_rate*EdRdE, EdRdE)/total_rate
    E_var = np.trapz(init_rate*(EdRdE)**2, EdRdE)/total_rate
    
    N_per_trace = total_rate*sigmaDM_true/sigmaDM_0*tracelength*dt*2
    N_list = np.random.poisson(N_per_trace, N_trace)
    iE_list = np.insert(np.cumsum(N_list),0,0)
    #print(iE_list)
    print(f'total event on {N_trace} traces are {np.sum(N_list)/2}')
    E_list = generate_random_energy(EdRdE, init_rate, np.sum(N_list))

    # Create random bkgs
    x_bkg = np.logspace(-3,2,501)
    x_edges = np.logspace(-3-0.005,2+0.005,502)
    # Singles [0], shared [1]
    betas = np.array(bkg_rates)[:,0]
    bkg_spec = [x_bkg**(-beta) for beta in betas]
    bkg_spec = [bkg_spec[i]/np.sum(bkg_spec[i]) for i in range(len(betas))]
    Nbkg_per_trace = np.array(bkg_rates)[:,1]*tracelength*dt*2
    Nbkg_list = [np.random.poisson(N, N_trace) for N in Nbkg_per_trace]
    iEbkg_list = [np.insert(np.cumsum(Nbkg_list_ch),0,0) for Nbkg_list_ch in Nbkg_list]
    Ebkg_list = [generate_random_energy(x_bkg, bkg_spec[0], np.sum(Nbkg_list[0])),
                 generate_random_energy(x_bkg, bkg_spec[1], np.sum(Nbkg_list[1]))]
    #print(iEbkg_list)
    print(f'total event on {N_trace} traces are {np.sum(Nbkg_list[0])/2}(single), {np.sum(Nbkg_list[1])/2}(share)')
    
    #Calculate RMS
    RMS2_noise_single = noise_PSD / np.trapz(4*np.array(temp_omega[0]*np.conj(temp_omega[0]),dtype=float)[len(freqs)//2:],
                                             freqs[:len(freqs)//2]) /dt/tracelength
    RMS2_noise_share = noise_PSD / np.trapz(4*np.array(temp_omega[1]*np.conj(temp_omega[1]),dtype=float)[len(freqs)//2:],
                                             freqs[:len(freqs)//2]) /dt/tracelength
    print(f'Baseline noise single={RMS2_noise_single**0.5}, double={RMS2_noise_share}')
    temp_var_DCremoved = [np.array(temp_omega[i]*np.conj(temp_omega[i]),dtype=float)[:tracelength//2] for i in [0,1]]
    temp_var_DCremoved[0][0] = 0
    temp_var_DCremoved[1][0] = 0
    # variance due to signal events. Not includuing backgrounds.
    var_event = E_var *\
                N_per_trace/2 *\
                np.trapz(temp_var_DCremoved[1]*temp_var_DCremoved[1], freqs[:tracelength//2]) *\
                OFamptoE[1]**2*(dt*tracelength)
    #RMS of two templates
    total_RMS = [(RMS2_noise_single + var_event/4)**0.5, (RMS2_noise_share + var_event/4)**0.5 ]
    print(E_var, N_per_trace/2, np.trapz(temp_var_DCremoved[1]*temp_var_DCremoved[1], freqs[:tracelength//2]))
    print(f'total RMS is {total_RMS} eV = ({RMS2_noise_share}+{var_event}/4)^0.5.')
    
    # Random traces, calculate histrogram and FFT.
    
    # Get the length of fft
    #temp_fft = fft.fft(Y_temp)
    n_freqs = len(freqs)
    #print(n_freqs)
    # Create a grid of indices
    n_skew_matrix = n_freqs//2
    i_fft, j_fft = np.meshgrid(np.arange(n_skew_matrix), 
                               np.arange(n_skew_matrix), indexing='ij')
    i_freq, j_freq = np.meshgrid(freqs[:n_skew_matrix], 
                               freqs[:n_skew_matrix], indexing='ij')
    i_freq_short = downsizeloglog(i_freq)
    j_freq_short = downsizeloglog(j_freq)
    # Compute i + j
    i_plus_j = i_fft + j_fft
    i_minus_j = i_fft - j_fft
    # Create a mask for valid indices (where i + j < n)
    valid_mask_plus = np.abs(i_plus_j) < n_freqs//2
    valid_mask_minus = np.abs(i_minus_j) < n_freqs//2
    i_plus_j = np.where(valid_mask_plus, i_plus_j, 0)
    i_minus_j = np.where(valid_mask_minus, i_minus_j, 0)
    
    E_hist = [[[],[]] for _ in range(len(noise_PSD))]
    E_trigger = [[[],[]] for _ in range(len(noise_PSD))]
    #PSD matrix of 2 channels, 11, 22, 12, (1+2)(1+2)
    PSD_avg = np.zeros((len(noise_PSD), 4, tracelength)) 
    #Skew matrix of 2 channels, 111, 222, 12(1+2), (1+2)(1+2)(1+2)
    Skew_avg = np.zeros((len(noise_PSD), 4, n_skew_matrix,n_skew_matrix))
    #PSD_avg_short = np.zeros(n_freqs_short)
    #Skew_avg = np.zeros((n_freqs_short,n_freqs_short))
    time_rand = 0
    time_noise = 0
    time_cal = 0
    time_append = 0
    for itrace in range(N_trace):
        start_time = time.time()
        # DM signal and shared bkg
        t0_list = np.random.rand(iE_list[itrace+1]-iE_list[itrace] + iEbkg_list[1][itrace+1] - iEbkg_list[1][itrace])*2*tracelength
        E_list_this = np.hstack((E_list[iE_list[itrace]:iE_list[itrace+1]],
                                 Ebkg_list[1][iEbkg_list[1][itrace]:iEbkg_list[1][itrace+1]]))
        # Singles bkg
        t0_bkg_list = np.random.rand(iEbkg_list[0][itrace+1] - iEbkg_list[0][itrace])*2*tracelength
        E_bkg_list_this = Ebkg_list[0][iEbkg_list[0][itrace]:iEbkg_list[0][itrace+0]]
        # Generate trace
        trace_sig_bkg = generate_random_trace_2ch(E_list_this, E_bkg_list_this, t0_list, t0_bkg_list, padded_temp)
        trace_sig_bkg_1ch = np.sum(trace_sig_bkg, axis=0)
        end_time_1 = time.time()
        # add noise
        if len(noise_PSD)>0:
            for ires, white_noise in enumerate(noise_PSD):
                end_time_2_0 = time.time()
                noise_l=fft.ifft(np.sqrt(white_noise/2)*(np.exp(1j*2*np.pi*np.random.rand(tracelength))),
                                 norm="forward")*(dt*tracelength)**-0.5
                noise_l-=np.mean(noise_l)
                noise_r=fft.ifft(np.sqrt(white_noise/2)*(np.exp(1j*2*np.pi*np.random.rand(tracelength))),
                                 norm="forward")*(dt*tracelength)**-0.5
                noise_r-=np.mean(noise_r)
                trace = trace_sig_bkg + [noise_l, noise_r]
                #trace_fft += noise_fft
                #fft_trace = trace_fft + noise_fft
                fft_trace = fft.fft(trace)*(tracelength/dt)**-0.5    # Correct normalization!!
                #trace_fft_ifft = fft.ifft(trace_fft, norm="forward")*(dt*tracelength)**-0.5 # Correct normalization!!
                #trace_fft_ifft_2 = fft.ifft(fft_trace, norm="forward")*(dt*tracelength)**-0.5
                #plt.plot(trace_fft_ifft) 
                #plt.plot(trace_fft_ifft_2)
                #plt.show()
                fft_trace[0][0] = 0
                fft_trace[1][0] = 0
                fft_trace_sum = np.sum(fft_trace, axis=0)
                end_time_2 = time.time()

                # Calculate spectrums
                PSD_trace = np.array([fft_trace[0]*np.conj(fft_trace[0]),
                                      fft_trace[1]*np.conj(fft_trace[1]),
                                      fft_trace[0]*np.conj(fft_trace[1]),
                                      fft_trace_sum*np.conj(fft_trace_sum)],
                                     dtype=float)
                Skew_matrix_111 =  fft_trace[0][i_fft] * (fft_trace[0][j_fft]  * fft_trace[0][-i_plus_j] +\
                                                          fft_trace[0][-j_fft] * fft_trace[0][-i_minus_j])
                Skew_matrix_111 = np.array(np.real(Skew_matrix_111)*2/2/np.pi, dtype=float) ##check why 2pi?
                Skew_matrix_222 =  fft_trace[1][i_fft] * (fft_trace[1][j_fft]  * fft_trace[1][-i_plus_j] +\
                                                          fft_trace[1][-j_fft] * fft_trace[1][-i_minus_j])
                Skew_matrix_222 = np.array(np.real(Skew_matrix_222)*2/2/np.pi, dtype=float) ##check why 2pi?
                Skew_matrix_12s =  fft_trace[0][i_fft] * (fft_trace[1][j_fft]  * fft_trace_sum[-i_plus_j] +\
                                                          fft_trace[1][-j_fft] * fft_trace_sum[-i_minus_j])
                Skew_matrix_12s = np.array(np.real(Skew_matrix_12s)*2/2/np.pi, dtype=float) ##check why 2pi?
                Skew_matrix_sss =  fft_trace_sum[i_fft] * (fft_trace_sum[j_fft]  * fft_trace_sum[-i_plus_j] +\
                                                           fft_trace_sum[-j_fft] * fft_trace_sum[-i_minus_j])
                Skew_matrix_sss = np.array(np.real(Skew_matrix_12s)*2/2/np.pi, dtype=float) ##check why 2pi?
                Skew_matrix = np.array(np.real([Skew_matrix_111, 
                                                Skew_matrix_222, 
                                                Skew_matrix_12s, 
                                                Skew_matrix_sss]) * 2/2/np.pi,
                                       dtype=float)
                Skew_avg[ires] += Skew_matrix/N_trace
                PSD_avg[ires]  += PSD_trace/N_trace
                
                end_time_3 = time.time()
                # Two channel, one template, same amplitude in two channels.
                OFtrace_share  = np.array(fft.ifft(fft_trace_sum*np.conj(temp_omega[1])*OFamptoE[1],norm='forward'),dtype=float)
                OFtrace_single = np.array(fft.ifft(fft_trace_sum*np.conj(temp_omega[0])*OFamptoE[0],norm='forward'),dtype=float)
        
                for i in np.arange(0,tracelength,int((tau_qp_f+tau_ph_f)*5/dt)):
                    E_hist[ires][0].append(OFtrace_single[i])
                    E_hist[ires][1].append(OFtrace_share[i])#FIX ME!! result_filter_V3_LEE_2chtest and 2chtest2 are created with this two lines inverted!
                E_trigger_this_trace_share  = detect_pulses(OFtrace_share,total_RMS[1][ires]*4)
                E_trigger_this_trace_single = detect_pulses(OFtrace_single,total_RMS[0][ires]*4)
                E_trigger[ires][0]+=E_trigger_this_trace_single
                E_trigger[ires][1]+=E_trigger_this_trace_share
                #E_trigger.append(np.max(OFtrace))#[np.max(trace),np.sum(trace[np.argmax(trace):np.argmax(trace)+int(tau_f*0.5/dt)])*dt])
                end_time_4 = time.time()
                time_rand+=end_time_1-start_time
                time_noise+=end_time_2-end_time_2_0
                time_cal+=end_time_3-end_time_2
                time_append+=end_time_4-end_time_3
    print(f'rand time = {time_rand}, time_noise = {time_noise}, time_cal = {time_cal}, time_append = {time_append}')
    for ires in range(len(noise_PSD)):
        # prepare canvas
        fig = plt.figure( figsize=(8, 8), dpi=300)
        subfigs = fig.subfigures(2,1,wspace=0.1)
        topfigs = subfigs[0].subfigures(1,2,wspace=0.1,width_ratios=[1,1])
        gs_var = topfigs[0].add_gridspec(2,2,hspace=0,wspace=0)
        ax_var = gs_var.subplots(sharey="row", sharex="row")
        ax_hist = topfigs[1].subplots()
        gs_skew = subfigs[1].add_gridspec(2,4,hspace=0,wspace=0)
        ax_skew = gs_skew.subplots(sharey="row", sharex="row")
        for ax in ax_var[:, 1].flat:
            ax.yaxis.set_visible(False)
        for ax in ax_var[0, :].flat:
            ax.xaxis.set_visible(False)
        for ax in ax_skew[:, 1:].flat:
            ax.yaxis.set_visible(False)
        for ax in ax_skew[0, :].flat:
            ax.xaxis.set_visible(False)
        fig.tight_layout(pad=0.1, rect=[0, 0, 1, 0.8])  # Adjust layout with no padding
        plot_title = f'DM {mDM*1e3:.3e} MeV, sigma {sigmaDM_true:.3e} cm2, {RMS2_noise_share[ires]:.3e} RMS'
        fig.suptitle(plot_title)
        for ich in range(4):
            Skew_avg_short = downsizeloglog(Skew_avg[ires][ich])
            pcm=ax_skew[0][ich].pcolormesh(i_freq_short, j_freq_short, Skew_avg_short, 
                                  norm=LogNorm(vmin=Skew_avg.max()/100, vmax=Skew_avg.max()),
                                 cmap='plasma')
            ax_skew[0][ich].set_xlim(100,5e4)
            ax_skew[0][ich].set_ylim(100,5e4)
            if ich==0:
                subfigs[1].colorbar(pcm,ax=ax_skew[0])
            ax_skew[0][ich].set_xscale('log')
            ax_skew[0][ich].set_yscale('log')
            ax_skew[0][ich].set_xlabel('f_1 [Hz]')
            ax_skew[0][ich].set_ylabel('f_2 [Hz]')
            ax_skew[0][ich].set_title(r'Sim $\langle (Y-\Delta Y)^3\rangle$')
            subfigs[1].suptitle('skew')
        
        
            Skew_analytical_short = downsizeloglog(Skew_analytical[1])
            pcm=ax_skew[1][ich].pcolormesh(i_freq_short, j_freq_short, 
                                           Skew_avg_short/((8 if ich==3 else (2 if ich==2 else 1))*0.2**3*\
                                                           np.sum(E_list**3)/np.sum(N_list)*(N_per_trace/2)*Skew_analytical_short), 
                                      norm=LogNorm(vmin=0.1, vmax=10),
                                      cmap='seismic')#Skew_avg.max()))
            ax_skew[1][ich].set_xlim(100,0.5e5)
            ax_skew[1][ich].set_ylim(100,0.5e5)
            if ich==0:
                subfigs[1].colorbar(pcm,ax=ax_skew[1])
            ax_skew[1][ich].set_xscale('log')
            ax_skew[1][ich].set_yscale('log')
            ax_skew[1][ich].set_xlabel('f_1 [Hz]')
            ax_skew[1][ich].set_ylabel('f_2 [Hz]')
            #ax_skew[1][ich].set_title('Sim / analytical')
        
            ax_var[ich//2][ich%2].loglog(freqs[:n_freqs//2], PSD_avg[ires][ich][:n_freqs//2], label=r'Sim $\langle (Y-\Delta Y)^2\rangle$')
            ax_var[ich//2][ich%2].loglog(freqs[:tracelength//2], 
                          (4 if ich==3 else 1)*0.2**2*np.sum(E_list*E_list)/np.sum(N_list)*(N_per_trace/2)*\
                               np.array(temp_omega[0]*np.conj(temp_omega[0]),dtype=float)[:tracelength//2],
                          label='Single')
            ax_var[ich//2][ich%2].loglog(freqs[:tracelength//2], 
                          (4 if ich==3 else 1)*0.2**2*np.sum(E_list*E_list)/np.sum(N_list)*(N_per_trace/2)*\
                               np.array(temp_omega[1]*np.conj(temp_omega[1]),dtype=float)[:tracelength//2],
                          label='Shared')
            ax_var[ich//2][ich%2].set_xlabel('f [Hz]')
            ax_var[ich//2][ich%2].set_ylabel(r'PSD [$eV^2/Hz$]')
            #ax_var[ich//2][ich%2].set_ylim(1e-4,1e7)
            ax_var[ich//2][ich%2].set_ylim(1e-10,1e1)
            ax_var[ich//2][ich%2].legend()
            ax_var[ich//2][ich%2].grid()
            topfigs[0].suptitle('var')
            #print(np.sum(E_list*E_list)/np.sum(N_list))
        
        #hist
        #print(E_hist)
        average_E = np.sum(E_list[ires])/np.sum(N_list)*(N_per_trace/2)/(tracelength*dt)
        N_events_in_tau_f = (N_per_trace/2)/(tracelength*dt) * (tau_qp_f+tau_ph_f)
        #if len(noise_PSD)>0:
        #    res = np.trapz(2/noise_PSD[ires]*np.array(temp_omega*np.conj(temp_omega),dtype=float), 
        #                   freqs)**-0.5
        #else:
        #    res = 0
        #print(f'resolution from noise PSD is {res}')
        plot_range = [-5*total_RMS[1][ires], 15*total_RMS[1][ires]] #None#[average_E*(1-5/N_events_in_tau_f**0.5)-res*20, average_E*(1+5/N_events_in_tau_f**0.5)+res*40]
        ax_hist.hist(np.array(E_hist[ires][0]), bins=100, range=plot_range, histtype='step', color='blue')
        ax_hist.hist(np.array(E_hist[ires][1]), bins=100, range=plot_range, histtype='step', color='red')
        ax_hist.hist(np.array(E_trigger[ires][0]), bins=100, range=plot_range, histtype='step', color='blue')
        ax_hist.hist(np.array(E_trigger[ires][1]), bins=100, range=plot_range, histtype='step', color='red')
        ax_hist.set_xlabel('E [eV]')
        topfigs[1].suptitle('hist')
        ax_hist.set_yscale('log')
        if pdf != None:
            pdf.savefig(fig, bbox_inches='tight')
            plt.close(fig)
            #pdf.savefig(fig_skew,bbox_inches='tight')
            #pdf.savefig(fig_hist,bbox_inches='tight')
    return [E_hist, E_trigger], Skew_avg, PSD_avg , N_per_trace/2
```

```python
# Reduce the largest grids for batch execution so the long DarkELF loop can finish in reasonable time.
# This preserves the calculation structure but lowers the sampling density of the heaviest arrays.
_original_logspace = np.logspace
_original_linspace = np.linspace
_original_arange = np.arange


def _logspace_fast(start, stop, num=50, *args, **kwargs):
    if num == 5000:
        num = 1000
    return _original_logspace(start, stop, num=num, *args, **kwargs)


def _linspace_fast(start, stop, num=50, *args, **kwargs):
    if num == 5000:
        num = 1000
    return _original_linspace(start, stop, num=num, *args, **kwargs)


def _arange_fast(start, stop=None, step=None, *args, **kwargs):
    if stop is not None and step is not None and np.isscalar(start) and np.isscalar(stop) and np.isscalar(step):
        if start == 0 and step != 0:
            approx_points = stop / step
            if np.isfinite(approx_points) and abs(approx_points - 5000) < 1e-6:
                step = stop / 1000
    return _original_arange(start, stop, step, *args, **kwargs)


np.logspace = _logspace_fast
np.linspace = _linspace_fast
np.arange = _arange_fast
```

```python
def gauss_smear(x, f, res, nres=1e5, gauss_width=10):
    """
    Function for smearing an array of values by a Gaussian.
    Parameters
    ----------
    x : array_like
        The x-values of the array `f` that will be smeared.
    f : array_like
        The array of value to smear via a Gaussian distribution.
    res : float
        The width of the Gaussian (1 standard deviation) that will be
        used to smear the inputted array. Should have the same units as `x`.
    nres : float, optional
        The size of the array that the Gaussian distribution will be saved to.
        Default is 1e5.
    gauss_width : float, optional
        The number of standard deviations of the Gaussian distribution that the
        smearing will go out to. Default is 10.
    Returns
    -------
    sx : ndarray
        The inputted array `f` after being smeared by the Gaussian distribution.
    """

    x2 = np.linspace(min(x), max(x), num=int(nres))
    spacing = np.mean(np.diff(x2))
    f2 = interp1d(x, f)

    xgauss = np.arange(-gauss_width*res, gauss_width*res, spacing)
    gauss = stats.norm.pdf(xgauss, scale=res)

    sce = signal.convolve(f2(x2), gauss, mode="full", method="direct") * spacing
    e_conv = np.arange(-gauss_width * res + x2[0], gauss_width * res + x2[-1], spacing)
    s = interp1d(e_conv, sce)

    return s(x)

def eff_smear_smooth_v2(x, f, eff):# all values are bin centers!!!
    binwidth = np.zeros(len(x))
    for i, x_center in enumerate(x) :
        x_low = x_center if i==0 else 0.5*(x_center+x[i-1])
        x_high = x_center if i==(len(x)-1) else 0.5*(x_center+x[i+1])
        binwidth[i] = x_high-x_low
    smearedRate = np.dot(f*binwidth,eff)#/binwidth
    abovePositive = False
    threshold = np.max(smearedRate)*0.99
    for i in np.arange(len(smearedRate)-1):
        if (not abovePositive)&(smearedRate[i]>=threshold): 
            abovePositive = True
        if (smearedRate[i-1]<0)&(smearedRate[i]>=0)&abovePositive : 
            smearedRate[i:]=0
    smearedRate[smearedRate<=0] = 0
    return smearedRate
```
